# Support Vector Machines (SVM)

Support Vector Machine is a powerful supervised algorithm used for classification, regression, and outlier detection.

**How It Works:**

SVM finds an optimal boundary (hyperplane) in a high-dimensional space that maximizes the margin, the distance between the boundary and the nearest data points of any class (support vectors).

* **Classification (SVC):** Separates data classes by placing a decision boundary with maximum margin.
* **Regression (SVR):** Fits a margin boundary (epsilon tube) that encompasses as many data points as possible while minimizing error.

**Best Use Cases:**

* Medium-sized datasets with clear margins of separation.
* High-dimensional datasets, such as text classification, bioinformatics, and image recognition.
