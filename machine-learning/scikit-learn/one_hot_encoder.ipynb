One-Hot Encoding is a technique used to convert categorical data (text labels or categories) into a binary numerical format so machine learning algorithms can process it.

How It Works
It creates a new binary column (containing 0 or 1) for every unique category present in the dataset. For each data row, a 1 is assigned to the column representing its category, and 0 is assigned to all other columns.

Example:
Converting a "Color" column with values Red, Green, and Blue:

```python
import pandas as pd

d = {'sales': [100000,222000,1000000,522000,111111,222222,1111111,20000,75000,90000,1000000,10000],
'city': ['Tampa','Tampa','Orlando','Jacksonville','Miami','Jacksonville','Miami','Miami','Orlando','Orlando','Orlando','Orlando'],
'size': ['Small', 'Medium','Large','Large','Small','Medium','Large','Small','Medium','Medium','Medium','Small',]}
```

```python
df_ohe = pd.DataFrame(data = d)

df_ohe.head()
```

```python
df_ohe['city'].unique
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False).set_output(transform='pandas')
```

```python
ohetransform = ohe.fit_transform(df_ohe[['city']])
```

```python
ohetransform
```

```python
df_ohe = pd.concat([df_ohe, ohetransform], axis=1)

df_ohe = df_ohe.drop(columns=['city'])
```
