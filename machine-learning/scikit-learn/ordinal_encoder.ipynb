Ordinal Encoding converts ordered categorical text values into sequential integers (0, 1, 2, ...) while preserving their natural rank or order.

**How It Works:**

Unlike One-Hot Encoding, Ordinal Encoding keeps everything in a single column by mapping categories according to a defined hierarchy.

**Best Use Cases:**

Categorical variables with an inherent sequence (e.g., T-shirt sizes [S, M, L, XL], customer ratings [Poor, Average, Good]).

```python
import pandas as pd

d_oe = {'sales': [100000,222000,1000000,522000,111111,222222,1111111,20000,75000,90000,1000000,10000],
'city': ['Tampa','Tampa','Orlando','Jacksonville','Miami','Jacksonville','Miami','Miami','Orlando','Orlando','Orlando','Orlando'],
'size': ['Small', 'Medium','Large','Large','Small','Medium','Large','Small','Medium','Medium','Medium','Small',]}

df_oe = pd.DataFrame(data=d_oe)

df_oe.head()
```

```python
df_oe['size'].unique()
```

```python
sizes = ['Small', 'Medium', 'Large']
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[sizes])
```

```python
oe.fit_transform(df_oe[['size']])
```

```python
oe.fit_transform(df_oe[['size']])
```

```python
df_oe['size'] = oe.fit_transform(df_oe[['size']])

df_oe.head()
```
