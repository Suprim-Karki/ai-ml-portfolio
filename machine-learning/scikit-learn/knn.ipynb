# K-Nearest Neighbors (KNN)

K-Nearest Neighbors (KNN) is a non-parametric, instance-based algorithm used for classification, regression, and missing value imputation (KNNImputer).

**How It Works:**

KNN operates on a spatial principle: "things that are close to each other are similar." 

For any new query point, it finds the K closest data points in feature space using a distance metric.

Classification : Predicts the majority class among K neighbors.

Regression     : Predicts the average target value among K neighbors.

Imputation     : Fills missing feature values using the average of K nearest rows.


**Formula / Mathematical Concept:**

Euclidean Distance between two points $p$ and $q$ across $n$ features:

distance = sqrt( sum( (p_i - q_i)^2 ) )

Where:

p_i, q_i: Feature values for data point $p$ and data point $q$
    
K: User-defined number of neighbors to evaluate

**Best Use Cases:**

Small-to-medium datasets where feature relationships are complex or non-linear.

Imputing missing values based on similarity across other complete features (KNNImputer).
