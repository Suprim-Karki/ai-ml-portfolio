# Simple Imputer (Handle Missing Data)

Simple Imputer fills in missing values (NaN, None, or placeholders) in numerical or categorical columns using basic statistical strategies.

**How It Works:**

It calculates a summary statistic on non-missing data during fit() and replaces every NaN with that value during transform().

**Strategies & Concepts:**

mean: x_imputed = mean(column) (Numerical only)

median: x_imputed = median(column) (Numerical only, robust to outliers)

most_frequent: x_imputed = mode(column) (Categorical or numerical)

constant: x_imputed = fill_value (e.g., replace missing with -1 or "Missing")

**Best Use Cases:**

Baseline preprocessing for handling incomplete datasets before sending data into algorithms that cannot handle missing values.

```python
import pandas as pd
import numpy as np

miles = pd.DataFrame({'farthest_run_mi':[50,62,np.nan,100,26,13,31,50]})
```

```python
miles.isna()
```

```python
from sklearn.impute import SimpleImputer

imp_mean = SimpleImputer(strategy='mean')

imp_mean.fit_transform(miles)
```

```python
imp_median = SimpleImputer(strategy='median')

imp_median.fit_transform(miles)
```

```python
imp_mode = SimpleImputer(strategy='most_frequent')

imp_mode.fit_transform(miles)
```

```python
imp_constant = SimpleImputer(strategy='constant', fill_value=13)

imp_constant.fit_transform(miles)
```

```python
names = pd.DataFrame({'names':['messi','ronaldo','pele','maradona',np.nan,'neymar','kdb']})

names.isna()
```

```python
imp_constant_category = SimpleImputer(strategy='constant', fill_value = 'lewandowiski')

imp_constant_category.fit_transform(names)
```

**To show which rows have been imputed:**

```python
imp_mean_marked = SimpleImputer(add_indicator=True)

imp_mean_marked.fit_transform(miles)
```
