# Random Forest Algorithm

Random Forest is an ensemble learning method that combines multiple individual Decision Trees to produce a more accurate and stable prediction.

**How It Works:**

It builds a "forest" of decision trees trained on random subsets of the data (bootstrapping) and random subsets of features (feature bagging).

* **Classification:** Aggregates predictions across all decision trees using a majority vote.
* **Regression:** Aggregates predictions by computing the average output across all individual decision trees.

**Formula / Mathematical Concept:**

Final prediction aggregation across $N$ trees:

$$\hat{y} = \frac{1}{N} \sum_{i=1}^{N} T_i(x)$$

Where:

* $N$: Total number of decision trees in the forest
* $T_i(x)$: Output prediction from the $i$-th individual decision tree

**Best Use Cases:**

* High-dimensional tabular datasets where high predictive accuracy is preferred over model interpretability.
* Complex tabular problems prone to overfitting when using a single decision tree.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv('500hits.csv' , encoding='latin-1')

df = df.drop(columns=["PLAYER","CS"])
```

```python
X = df.iloc[:, 0:13]
y = df.iloc[:,13]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13)
```

```python
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier()

rfc.get_params()
```

```python
rfc.fit(X_train, y_train)

y_pred = rfc.predict(X_test)

rfc.score(X_test, y_test)
```
