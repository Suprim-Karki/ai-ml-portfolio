# Standardization

Standardization rescales data so it has a mean of 0 and a standard deviation of 1. 

It does not bind values to a fixed range.

Formula: $$x_{\text{new}} = \frac{x - \mu}{\sigma}$$

Key Characteristic: Preserves outliers and works best when the data follows a Normal (Gaussian) distribution.

Common Use Cases: Algorithms that assume normally distributed data or rely on distance/variance (e.g., Logistic Regression, Support Vector Machines, Principal Component Analysis).

```python
import pandas as pd
df = pd.read_csv('500hits.csv', encoding='latin-1')
```

```python
df = df.drop(columns=['PLAYER', 'CS'])
```

```python
df.describe().round(2)
```

```python
X1 = df.iloc[:, 0:13].values
X2 = df.iloc[:, 0:13].values
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X1 = scaler.fit_transform(X1)
```

```python
X1 = pd.DataFrame(X1, columns=['YRS','G','AB','R','H','2B','3B'	,'HR','RBI','BB','SO','SB', 'BA'])
```

```python
X1.describe().round(3)
```

# Normalization

Normalization rescales data so that all values fit strictly within a fixed range, typically[0, 1]

Formula: $$x_{\text{new}} = \frac{x - x_{\text{min}}}{x_{\text{max}} - x_{\text{min}}}$$

Key Characteristic: Highly sensitive to outliers, as extreme values compress the normal data into a tiny range.

Common Use Cases: Algorithms that do not assume any specific distribution or require bounded inputs (e.g., K-Nearest Neighbors, Neural Networks, Image Processing pixel values).

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaleMinMax = MinMaxScaler(feature_range=(0,1))
```

```python
X2 = scaleMinMax.fit_transform(X2)

X2 = pd.DataFrame(X2, columns=['YRS','G','AB','R','H','2B','3B'	,'HR','RBI','BB','SO','SB', 'BA'])
```

```python
X2.describe().round(3)
```
